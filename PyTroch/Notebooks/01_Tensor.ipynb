## Data Types in PyTorch

| **Data Type** | **Dtype** | **Description** | **Common Use Cases** |
|---|---|---|---|
| **32-bit Floating Point** | `torch.float32` | Default floating-point dtype with a good balance between precision, speed, and memory usage. | Model weights, activations, loss calculations, general deep learning |
| **64-bit Floating Point** | `torch.float64` | Double-precision floating point with higher numerical accuracy and greater memory usage. | Scientific computing, numerical analysis, debugging |
| **16-bit Floating Point** | `torch.float16` | Half-precision floating point that reduces memory usage and can accelerate supported GPU operations. | Mixed-precision training, GPU inference |
| **BFloat16** | `torch.bfloat16` | 16-bit format with the same exponent range as `float32` but lower precision. | Large-model training, mixed precision, TPUs, supported CPUs and GPUs |
| **Float8 E4M3** | `torch.float8_e4m3fn` | 8-bit floating-point format optimized for relatively higher precision within float8 formats. | Experimental low-precision training and inference |
| **Float8 E5M2** | `torch.float8_e5m2` | 8-bit floating-point format with a wider range but lower precision than E4M3. | Experimental training, gradients, large-scale model workloads |
| **8-bit Signed Integer** | `torch.int8` | Signed integer with values from -128 to 127. | Quantized models, compact integer data |
| **16-bit Signed Integer** | `torch.int16` | Signed 16-bit integer for integer calculations requiring a larger range than `int8`. | Signal processing, compact numerical data |
| **32-bit Signed Integer** | `torch.int32` | Signed 32-bit integer. | Integer calculations, some indexing operations, interoperability |
| **64-bit Signed Integer** | `torch.int64` | Signed 64-bit integer, also known as `torch.long`. | Tensor indices, class labels, counters, integer calculations |
| **8-bit Unsigned Integer** | `torch.uint8` | Unsigned integer with values from 0 to 255. | Legacy image data, byte data, masks |
| **16-bit Unsigned Integer** | `torch.uint16` | Unsigned 16-bit integer. Availability and operator support depend on the PyTorch version. | Compact non-negative data, specialized applications |
| **32-bit Unsigned Integer** | `torch.uint32` | Unsigned 32-bit integer. Availability and operator support depend on the PyTorch version. | Specialized integer data and interoperability |
| **64-bit Unsigned Integer** | `torch.uint64` | Unsigned 64-bit integer. Availability and operator support depend on the PyTorch version. | Large non-negative integer values, specialized applications |
| **Boolean** | `torch.bool` | Stores `True` or `False` values. | Masks, comparisons, conditional selection |
| **Complex 64** | `torch.complex64` | Complex numbers with 32-bit real and imaginary components. | Signal processing, Fourier transforms, scientific computing |
| **Complex 128** | `torch.complex128` | Complex numbers with 64-bit real and imaginary components. | High-precision scientific and signal-processing calculations |
| **Quantized Signed Integer** | `torch.qint8` | Quantized signed 8-bit representation interpreted using a scale and zero point. | Quantized model weights and inference |
| **Quantized Unsigned Integer** | `torch.quint8` | Quantized unsigned 8-bit representation interpreted using a scale and zero point. | Quantized activations and inference |
| **Quantized 32-bit Integer** | `torch.qint32` | Quantized 32-bit signed integer representation. | Certain quantized layers and accumulators |





> **Note:** Dtype availability and operator support depend on the installed PyTorch version and hardware backend. Quantized dtypes such as `torch.qint8` and `torch.quint8` require quantization metadata, including a scale and zero point; they are not ordinary tensor dtypes for most `torch.tensor()` calls.

### 4. Mathematical Operations on tensors

#### 1. Scaler Operation

````python
x = torch.tensor([
    [1,2],
    [1,4]
])

# 1.Addition of a scalar to all elements in the tensor
y = x + 3
print(y)

# 2. subtraction of a scalar from all elements in the tensor
z = x - 3
print(z)

# 3. Multiplication of a scalar with all elements in the tensor
w = x * 3
print(w)

# 4. Division of a scalar by all elements in the tensor
v = x / 3
print(v)

# 5. Mode 
print(((x * 100)// 2)% 2)

# 6. Power
print(x**2)
````

#### 2. Element Wise Operation

````python
torch.manual_seed(45)
a = torch.rand(2,3)
b = torch.rand(2,3)

print(a)
print(b)
````

````python
# 1. Addition of two tensors
print("Addition : \n",a+b)

# 2. Subtraction of two tensors
print("Subtraction : \n",a-b)

# 3. Multiplication of two tensors
print("Multiplication : \n",a*b)

# 4. Division of two tensors
print("Division : \n",a/b)

# 5. Modulo operation between two tensors
print("Modulo Operation : \n",a%b)

# 6. Power operation between two tensors
print("Power Operation : \n",a**b)
````

#### 3. Operation on single tensors

````python
x = torch.tensor([-2,-1,0,1,2])
y = torch.tensor([1.0, 2.9, 3.5, 4.7, 5.1]) 

# 1. Absolute value of a tensor
print("Absolute Value : \n",torch.abs(x))

# 2. Square root of a tensor
print("Square Root : \n",torch.sqrt(x))

# 3. Exponential of a tensor
print("Exponential : \n",torch.exp(x))

# 4. Logarithm of a tensor
print("Logarithm : \n",torch.log(x))

# 5. Round off of a tensor
print("Round Off : \n",torch.round(x))

# 6. Negative of a tensor
print("Negative : \n",torch.neg(x))

# 7. Maximum value in a tensor
print("Maximum Value : \n",torch.max(x,y))

# 8. ceiling of a tensor
print("Ceiling : \n",torch.ceil(y))

# 9. floor of a tensor
print("Floor : \n",torch.floor(y))

# 10. clam of a tensor
print("Clamp : \n",torch.clamp(y,0,5)) # tensor, min, max

# 11. sigmoid for each element in a tensor
print("Sigmoid : \n",torch.sigmoid(y)) # sigmoid function

# 12. tanh for each element in a tensor
print("Tanh : \n",torch.tanh(y)) # tanh function

# 13. softmax for each element in a tensor
print("Softmax : \n",torch.softmax(y, dim=0)) # softmax function across row 

# 14. relu for each element in a tensor
print("ReLU : \n",torch.relu(y)) # relu function
````

### 6. Comparison Operation

````python
torch.manual_seed(45)
x = torch.randint(size=(2,4), low=0, high=5, dtype=torch.float32)
y = torch.randint(size=(2,4), low=0, high=5, dtype=torch.float32)

print("X: \n", x)
print("Y: \n",y)
````

### 7. Inplace Operation

````python
torch.manual_seed(45)

x = torch.randint(size=(2,4), low=0, high=5, dtype=torch.float32)
y = torch.randint(size=(2,4), low=0, high=5, dtype=torch.float32)

# 1.To add without making new tensor
print("Addition of X and Y: \n",x.add_(y), "\n") # Stores the new tensor to x

# 2. To multiply without making new tensor
print("Multiplication of X and Y: \n",x.mul_(y), "\n") # Stores the new tensor to x

# x.function_() represent inplace operation
````

#### 8. Copy a Tensor

````python
torch.manual_seed(45)
x = torch.randint(size=(2,4), low=0, high=5, dtype=torch.float32)

# If we assign this tensor to new variable they both we point to the same location in memory i.e change in one tensor will reflect on other as well
y = x
if id(x) == id(y):
    print("Both are pointing to same location in memory")
else:
    print("Both are not pointing to same location in memory")

# so to copy a tensor we use clone(), it creates a new tensor with the same data as the original tensor.
z = x.clone()
if id(x) == id(z):
    print("Both are pointing to same location in memory")
else:
    print("Both are not pointing to same location in memory")
````

#### 9. Tensor Operations on GPU

````python
if torch.backends.mps.is_available():
    
    print("GPU is available ! (via MPS)")
    device = torch.device("mps")
    
elif torch.cuda.is_available():
    
    print("GPU is available!! (via CUDA)")
    device = torch.device("cuda")
    print(f"Device Name : {torch.cuda.get_device_name()}")
    
else:
    
    print("No GPU found, using CPU")
    device = torch.device("cpu")

print(f"Active device: {device}")
print(torch.backends.mps.is_built())

````

### NOTE:  whenever  build tensors/models, move them to that device explicitly:

```
model = model.to(device)
x = x.to(device)
```

````python
# Creating a tensor on GPU
tensor_on_gpu = torch.tensor([1.0, 2.0, 3.0], device=device)
print(tensor_on_gpu)
````

````python
# Move Existing Tensors to GPU
torch.manual_seed(45)

# These Tensors are on CPU by default
x = torch.randint(size=(2, 3),low=0,high=100)
y = torch.randint(size=(2,3),low=0,high=100)

# Move tensors to GPU
x_gpu = x.to(device)
y_gpu = y.to(device)

print(f"Tensor X on GPU : {x_gpu}")
print(f"Tensor Y on GPU : {y_gpu}")

# Any Operation will be performed on the GPU
z = torch.add(x_gpu, y_gpu)
print(z)
````

### 10. Time difference b/w CPU and GPU computation

````python
import time
torch.manual_seed(45)

# Define Size of matrix
size = 20000

# Create a random matrix on CPU
x_cpu = torch.rand(size, size)
y_cpu = torch.rand(size, size)

# Time Measurement for CPU Operation
start_time = time.time()
result_cpu = torch.matmul(x_cpu, y_cpu)
cpu_time = time.time() - start_time

print(f"CPU Matrix Multiplication took {cpu_time:.4f} seconds")
````

````python
# Move to GPU
x_gpu = x_cpu.to(device)
y_gpu = y_cpu.to(device)

# Time Measurement for GPU Operation
start_time = time.time()
result_gpu = torch.matmul(x_gpu, y_gpu)
gpu_time = time.time() - start_time
print(f"GPU Matrix Multiplication took {gpu_time:.4f} seconds")


# Compare result
print(f"Speed Up: {cpu_time / gpu_time:.2f} times faster on GPU")
````
